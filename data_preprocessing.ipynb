```python
import fastf1
import pandas as pd
```

```python
import os

os.makedirs('./f1_cache', exist_ok=True)
fastf1.Cache.enable_cache('./f1_cache')
```

```python
years = [2020,2021,2022,2023,2024]
schedules=[]
for year in years:
    schedule = fastf1.get_event_schedule(year)
    schedules.append(schedule)

schedules=pd.concat(schedules)
schedules
```

```python
races = schedules[schedules['EventFormat'] == 'conventional']
races
```

```python
all_race_results = []
all_quali_results = []
for _, race in races.iterrows():
  gp_name = race['EventName']
  rnd = int(race['RoundNumber'])

  try:
    race_session = fastf1.get_session(year, rnd, 'R')
    race_session.load()

    race_results = race_session.results[['DriverNumber','Abbreviation','TeamName','Position','Points']]
    race_results['Year'] = race['EventDate'].year
    race_results['GrandPrix'] = gp_name
    race_results['Round'] = rnd

    all_race_results.append(race_results)
  except Exception as e:
      print(f'Unable to Fetch {gp_name} ({year}): {e}')


  try:
     q_session = fastf1.get_session(year, rnd, 'Q')
     q_session.load()
     q_results = q_session.results[['DriverNumber', 'Abbreviation','Position','Q1','Q2','Q3']]
     q_results['Year'] = race['EventDate'].year
     q_results['Round'] = rnd
     q_results['GrandPrix'] = gp_name

     all_quali_results.append(q_results)
  except Exception as e:
      print(f'Unable to Fetch Qualifying for {gp_name} ({year}): {e}')

race_df = pd.concat(all_race_results, ignore_index=True) if all_race_results else pd.DataFrame()
quali_df = pd.concat(all_quali_results, ignore_index=True) if all_quali_results else pd.DataFrame()

race_df.rename(columns={'Abbreviation': 'Driver'}, inplace=True)
quali_df.rename(columns={'Abbreviation': 'Driver'}, inplace=True)

df = pd.merge(
    race_df,
    quali_df[['Year', 'Round', 'GrandPrix', 'Driver', 'Position']].rename(columns={'Position': 'QualPosition'}),
    on=['Year', 'Round', 'GrandPrix', 'Driver'],
    how='left'
)
```

```python
df['Position'] = df['Position'].astype(int)
df['Points'] = df['Points'].astype(int)
df['QualPosition'] = df['QualPosition'].astype(int)
```

```python
df['Driver_AvgFinish_Season'] = df.groupby(['Year', 'Driver'])['Position'].transform('mean') #checking driver performance over the season
df['Team_AvgPoints_Season'] = df.groupby(['Year', 'TeamName'])['Points'].transform('mean') #checking team performance over the season
df['PrevRaceFinish'] = df.groupby('Driver')['Position'].shift(1).fillna(10)  #checking drivers short term past peprformance
df['Form_Last3'] = df.groupby('Driver')['Position'].transform(lambda x: x.rolling(3, min_periods=1).mean()) #to check driver form for the last 3 races
df
```

```python
# Average driver finishing position at specific GP (past years only)
df['AvgFinish_AtGP_PastYears'] = (
    df.groupby(['Driver', 'GrandPrix'])['Position'].transform(lambda x: x.shift().expanding().mean()))

# Average qualifying position at this GP (past years only)
df['AvgQual_AtGP_PastYears'] = (
    df.groupby(['Driver', 'GrandPrix'])['QualPosition'].transform(lambda x: x.shift().expanding().mean()))

```

```python
#scaling [0,1]
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaled = scaler.fit_transform(df[['Form_Last3', 'AvgFinish_AtGP_PastYears', 'AvgQual_AtGP_PastYears']].fillna(0))
df[['norm_Form_Last3', 'norm_AvgFinishAtGP', 'norm_AvgQualAtGP']]  = scaled
```

```python
#since lower finishing position is better in F1
df['norm_AvgFinishAtGP'] = 1 - df['norm_AvgFinishAtGP']  
df['norm_Form_Last3'] = 1 - df['norm_Form_Last3']
df['norm_AvgQualAtGP'] = 1 - df['norm_AvgQualAtGP']
df
```

```python
df['Drivers_Confidence'] = (
    0.4 * df['norm_Form_Last3'] +   #recent form matters most
    0.3 * df['norm_AvgFinishAtGP'] +
    0.3 * df['norm_AvgQualAtGP']
)
#(0.4+0.3+0.3=1) this is the weight of the factors considered
```

```python
# encoding categorical features
from sklearn.preprocessing import LabelEncoder

encoder_driver = LabelEncoder()
encoder_team = LabelEncoder()
encoder_gp = LabelEncoder()

df['DriverID'] = encoder_driver.fit_transform(df['Driver'])
df['TeamID'] = encoder_team.fit_transform(df['TeamName'])
df['GP_ID'] = encoder_gp.fit_transform(df['GrandPrix'])
```
